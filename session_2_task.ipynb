````python
ctc = float(input("Enter your CTC in lakhs: "))

# Calculate deductions
hra = ctc * 0.10
da = ctc * 0.05
pf = ctc * 0.03

# Calculate taxable salary after HRA, DA and PF deductions
salary = ctc - hra - da - pf

# Calculate tax based on CTC
if ctc < 5:
    tax = 0
elif ctc <= 10:
    tax = ctc * 0.10
elif ctc <= 20:
    tax = ctc * 0.20
else:
    tax = ctc * 0.30

# Annual in-hand salary
in_hand_annual = salary - tax

# Monthly in-hand salary
in_hand_monthly = in_hand_annual / 12

print("HRA:", hra, "Lakhs")
print("DA:", da, "Lakhs")
print("PF:", pf, "Lakhs")
print("Tax:", tax, "Lakhs")
print("Annual In-hand Salary:", in_hand_annual, "Lakhs")
print("Monthly In-hand Salary:", in_hand_monthly, "Lakhs")
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
# Write code here
cost_price = float(input("Enter your actual cost on the product: "))
selling_price = float(input("Enter your selling price on the product: "))

print("Your product has cost price:", cost_price,
      "and selling price:", selling_price)

if selling_price > cost_price:
    profit = selling_price - cost_price
    print("Yes, you are in profit.")
    print("Your profit is:", profit)

elif selling_price < cost_price:
    loss = cost_price - selling_price
    print("You are in loss.")
    print("Your loss is:", loss)

else:
    print("There is no profit and no loss.")
````

### `Problem 5` - Exercise 12: Display Fibonacci series up to 10 terms.

*Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two numbers before it. The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34*

````python
# Write code here
x = 10

# Initial
fib1 = 0
fib2 = 1

# Display the first two Fibonacci numbers
print("fib of 1 =", fib1)
print("fib of 2 =", fib2)

# Generate Fibonacci sequence up to x terms
for i in range(3, x + 1):
    # Calculate the next Fibonacci number
    next_fib = fib1 + fib2
    # Update fib1 and fib2 for the next iteration
    fib1 = fib2
    fib2 = next_fib
    # Display the Fibonacci number
    print("fib of", i, "=", next_fib)
````

### `Problem 7` - Reverse a given integer number.

Example:

`Input:`

```bash
76542
```

`Output:`

```bash
24567
```

````python
# Write code here
print("I am reversing the number 76542")

x = 76542
reverse = 0

while x > 0:

    digit = x % 10
    reverse = (reverse * 10) + digit
    # Remove the last digit from x
    x = x // 10

print("The reversed number is:", reverse)
````

###`Problem 10`: Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number. The numbers obtained should be printed in a space-separated sequence on a single line.



````python
# Write code here
print("Code for the No. between 1000 to 3000 ")


result = []

for num in range(1000, 3001):
    all_even = True
    for digit in str(num):
        # Check if the digit is even
        if int(digit) % 2 != 0:
            all_even = False
            break
    # If all digits are even, add the number to the result list
    if all_even:
        result.append(str(num))

# Print the numbers in a space-separated sequence on a single line
print(" ".join(result))
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
# Write code here
import math

x = 0
y = 0

while True:
    movement = input()

    if movement == "!":
        break

    direction, steps = movement.split()
    steps = int(steps)

    if direction == "UP":
        y += steps
    elif direction == "DOWN":
        y -= steps
    elif direction == "LEFT":
        x -= steps
    elif direction == "RIGHT":
        x += steps

# Calculate distance from the original point (0, 0)
distance = math.sqrt(x ** 2 + y ** 2)

# Print nearest integer
print(round(distance))
````
